```python
import pandas as pd, matplotlib.pyplot as plt
from IPython.display import display, Markdown

df = pd.DataFrame([
    {"Article":"ISRO","length":45000,"sections":18,"images":12,"infobox":15},
    {"Article":"Chandrayaan-3","length":28000,"sections":12,"images":8,"infobox":10},
    {"Article":"Small Village","length":1200,"sections":2,"images":0,"infobox":2},
])
df['Score'] = df['length']/1000*2 + df['sections']*5 + df['images']*8 + df['infobox']*4
df = df.sort_values('Score', ascending=False)
display(df)

plt.barh(df['Article'], df['Score'], color=['green','orange','red'])
plt.title('Weak Article Finder'); plt.show()
display(Markdown("**Score = length/1000*2 + sections*5 + images*8 + infobox*4** - Experimental richness, not official Wikipedia rating"))
```
