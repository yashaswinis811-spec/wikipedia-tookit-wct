```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

wiki_data = {
    "Indian Space Research Organisation": {
        "info": "ISRO is the space agency of India, headquartered in Bengaluru.",
        "People": ["Vikram Sarabhai", "A. P. J. Abdul Kalam"],
        "Organizations": ["NASA", "ESA"],
        "Places": ["Satish Dhawan Space Centre", "Bengaluru"],
        "Topics": ["Chandrayaan-3", "Mangalyaan"],
        "Related articles": ["India", "Space Research"]
    },
    "India": {
        "info": "India is a country in South Asia.",
        "People": ["A. P. J. Abdul Kalam"],
        "Organizations": ["ISRO"],
        "Places": ["Bengaluru", "New Delhi"],
        "Topics": ["Chandrayaan-3"],
        "Related articles": ["ISRO"]
    }
}

out = widgets.Output()

def show_article(topic):
    with out:
        clear_output()
        data = wiki_data[topic]
        display(Markdown(f"# {topic} - {data['info']}"))
        for cat in ["People", "Organizations", "Places", "Topics", "Related articles"]:
            display(Markdown(f"**{cat}:** {', '.join(data[cat])}"))

drop = widgets.Dropdown(options=list(wiki_data.keys()), value='Indian Space Research Organisation', description='Search:')
widgets.interactive_output(show_article, {'topic': drop})
display(drop, out)
```
